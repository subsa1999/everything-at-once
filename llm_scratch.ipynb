```python
# Read file
with open('data.txt', 'r', encoding='utf-8') as f:
  raw_text = f.read()
```

```python
# Tokenize the text
import re

tokens = [item.lower() for item in re.split(r'([.,;?()"\'\[\]]|-|--|\n|\s)', raw_text) if item.strip()]
len(tokens)
```

```python
# Creating Vocabulary
all_words = sorted(set(tokens))
all_words.extend(['<|endoftext|>', '<|unk|>'])
vocab = {token:index for index, token in enumerate(all_words)}
```

```python
class SimpleTokenizer:
  def __init__(self, vocab):
    self.str_to_int = vocab
    self.int_to_str = {i:s for s,i in vocab.items()}

  def encode(self, raw_text):
    pre = [item.lower() for item in re.split(r'([.,;?()"\'\[\]]|-|--|\n|\s)', raw_text) if item.strip()]
    pre = [item if item in self.str_to_int else '<|unk|>' for item in pre]
    return [self.str_to_int[s] for s in pre]

  def decode(self, ids):
    ## To be done: The output for 'Hey, how are you?' --> 'Hey ? how are you ?' --> because the tokens are joined
    return " ".join([self.int_to_str[id] for id in ids])
```

```python
st = SimpleTokenizer(vocab)
```

```python
st.decode(st.encode('Do you like tea or coffee?'))
```
